```python
import os
import sys

import mlflow

sys.path.append("/workspace")

from drsd.pipeline import SchemaDiscoveryPipeline

MLFLOW_TRACKING_URI = os.getenv("MLFLOW_TRACKING_URI")
assert MLFLOW_TRACKING_URI is not None
mlflow.set_tracking_uri(MLFLOW_TRACKING_URI)
```

```python
# MLflow
experiment_name = "proposal"
experiment = mlflow.get_experiment_by_name(experiment_name)
experiment_id = experiment.experiment_id if experiment is not None else mlflow.create_experiment(experiment_name)
```

```python
round_num = 32
for seed in [0, 1, 2]:
    for candidate_num in [256]:
        with mlflow.start_run(experiment_id=experiment_id, nested=True) as run:
            pipeline = SchemaDiscoveryPipeline(seed=seed, candidate_num=candidate_num)
            relationship_probabilities_by_subset, final_predictions_by_subset = None, None
            for round_id in range(1 + round_num):
                display(f"****************************** {round_id = } ******************************")
                relationship_probabilities_by_subset, final_predictions_by_subset = pipeline.run_round(
                    round_id,
                    relationship_probabilities_by_subset=relationship_probabilities_by_subset,
                    final_predictions_by_subset=final_predictions_by_subset,
                )
```
